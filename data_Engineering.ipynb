```python
print("Department of Electronic communication engineering !!! ")
elec_std = data[data['department'] == 'Electronics']
print(elec_std)
print("")
total_score = data['math_score'] + data['science_score'] + data['english_score'] + data['programming_score']
print("Average Score ",total_score.mean())
```

```python
print("Students with attendance percentage greater than 90%")
print(data[data['attendance_percentage']>90][['name','department', 'attendance_percentage']])
```

```python
print("Top performing students")
print(data[data['programming_score']>90] [['name','department', 'programming_score']])
```

```python
data['total_score'] = data['math_score'] + data['science_score'] + data['english_score'] + data['programming_score']
```

```python
print("Top performing students ")
print(data.sort_values(by='total_score',ascending = False)[['name','department','total_score']].head())
```

```python
dept_avg = data.groupby('department')['total_score'].mean().round(2)
print("==== Average department score ====")
print(dept_avg )
```

```python
dept_avg = data.groupby('department')['total_score'].mean().round(2)
print("==== Descending order Average department score ====")
avg = dept_avg.sort_values(ascending = False)
print(avg)
```

```python
dept_avg = data.groupby('department')['total_score'].mean().round(2)
print("==== Ascending Order Average department score ====")
avg_as = dept_avg.sort_values(ascending = True)
print(avg_as)
```
