```python
import matplotlib.pyplot as plt
import sqlite3
import pandas as pd

conn = sqlite3.connect('internship.db')
cursor = conn.cursor()

# Check if table exists
cursor.execute("SELECT name FROM sqlite_master WHERE type='table' AND name='internship';")
table_exists = cursor.fetchone()

if not table_exists:
    print("Error: The 'internship' table does not exist in the database. Please ensure it has been created.")
else:
    # Query to get average programming score by age
    chart_sql = """
    SELECT age, ROUND(AVG(programming_score), 2) AS avg_programming_score
    FROM internship
    GROUP BY age
    ORDER BY age ASC;
    """
    data = pd.read_sql(chart_sql, conn)

    if not data.empty:
        fig, ax = plt.subplots(figsize=(10, 6))
        plt.plot(data['age'], data['avg_programming_score'], marker='o', linestyle='-', color='purple')
        plt.title('Average Programming Score by Age')
        plt.xlabel('Age')
        plt.ylabel('Average Programming Score')
        plt.grid(True)
        plt.xticks(data['age'].unique()) # Ensure all ages are shown as ticks
        plt.show()
    else:
        print("No data available to plot a line chart for average programming score by age.")

conn.close()
```

```python
import os
def dir_size(path):
  if os.path.isfile(path):
    return os.path.getsize(path)/1024
  total = 0
  for dirpath,dirnames,filenames in os.walk(path):
    for f in filenames:
      total += os.path.getsize(os.path.join(dirpath,f))
  return total / 1024

csv_size = dir_size('/content/drive/MyDrive/Summer_Internship_2026/large_sales_data.csv')
parquet_size = dir_size('/content/sales_bronze.parquet')
reduction = (1 - parquet_size/csv_size)*100
print("CSV size : ",csv_size,"KB")
print("Parquet size : ",parquet_size,"KB")
print(f"Reduction : {reduction:.1f} % Smaller")
print(f"At 1 TB scale : 1000 GB -> parquet : {1000*(1-reduction/100):.0f} GB")
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder , StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score

print("All Libraries Imported Sucessfully !!!!")
```

```python
df = pd.read_csv("/content/drive/MyDrive/Summer_Internship_2026/student_performance.csv")
print("Data loaded : ",df.shape[0],"students",df.shape[1],"Columns")
print("Column Name : ",df.columns.to_list())
```

```python
print("=== Programming Score Statistics ===")
print(df['programming_score'].describe())
```

```python
print("=== GENDER ENCODING ===")
df_ml = df.copy()
le_gender = LabelEncoder()
df_ml['gender'] = le_gender.fit_transform(df_ml['gender'])
print(f"Gender encoding : {dict(zip(le_gender.classes_,le_gender.transform(le_gender.classes_)))}")
```

```python
print("=== DEPARTMENT ENCODING ===")
le_dept = LabelEncoder()
df_ml['department'] = le_dept.fit_transform(df_ml['department'])
print(f"Department Encoding : {dict(zip(le_dept.classes_,le_dept.transform(le_dept.classes_)))}")
```

```python
print("=== SEMESTER ENCODING ===")
le_sem = LabelEncoder()
df_ml['semester'] = le_sem.fit_transform(df_ml['semester'])
print(f"Semester Encoding : {dict(zip(le_sem.classes_,le_sem.transform(le_sem.classes_)))}")
```

```python
feature_col = [
    'math_score',
    'science_score',
    'attendance_percentage',
    'gender',
    'department'
]
x = df_ml[feature_col]
y = df_ml['programming_score']
print("Feature Matrix X shape : ",x.shape)
print("Target Vector Y shape : ",y.shape)
print(f"Target Range from  {y.min()} to {y.max()} mean : {y.mean():.1f}") # Corrected y.min and y.max to be functions
```

```python
X_train , X_test ,Y_train , Y_test = train_test_split(
    x,
    y,
    test_size = 0.2,
    random_state = 42
)
print(f"Total Students : {len(x)}")
print(f"Training Students : {len(X_train)} : ({len(X_train)/len(x)*100:.0f}) %")
print(f"Testing Students : {len(X_test)} ({len(X_test)/len(x)*100:.0f})")
print(f"Traininf target Range {Y_train.min()} to {Y_train.max()} mean : {Y_train.mean():.1f}")
print(f"Testing target Range {Y_test.min()} to {Y_test.max()} mean : {Y_test.mean():.1f}")
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
print("=== Features Scaled Sucessfully ===")
X_test_scaled =  scaler.transform(X_test)
print(f"Training feature mean (should be 0 ) : {X_train_scaled.mean():.4f}")
print("Traing feature Std (should be -1 ) : ",X_train_scaled.std())
```

```python
import numpy as np
lr_model = LinearRegression()
lr_model.fit(X_train_scaled,Y_train)
lr_pred = lr_model.predict(X_test_scaled)

lr_mae = mean_absolute_error(Y_test,lr_pred)
lr_mse = np.sqrt(mean_squared_error(Y_test,lr_pred))
lr_r2 = r2_score(Y_test,lr_pred)
lr_rmse = np.sqrt(lr_mse)

print("=== MODEL IN LINEAR REGRESSION ===")
print(f"MAE : {lr_mae:.2f}")
print(f"MSE : {lr_mse:.2f}")
print(f"R2 : {lr_r2:.4f}")
print(f"RMSE : {lr_rmse:.2f}")
```
