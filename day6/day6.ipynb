```python
print("=== MAKE MY FIRST CALL OF LLM MODEL ===")
print()
def ask_llm(user_message,system_message = "You are a helpful Assistant .",temperature = 0.7 , max_tokens = 500):
  """send the message to the llm and return the response text ."""
  response = client.chat.completions.create(
      model = MODEL,
      messages = [
          {"role":"system","content":system_message},
          {"role":"user","content":user_message}
      ],
      temperature = temperature,
      max_tokens = max_tokens
  )
  return response.choices[0].message.content

test_llm = ask_llm(
    "What is ETL in Data Engineering ? "
)
print(" === LLM RESPONSE ===")
print(test_llm)
```

```python
few_shot_prompt = """
Convert this employee text to JSON

Input: Ravi,50000,Chennai
Output: {"name":"Ravi","salary":"50000","city":"Chennai"}

Input: Priya,60000,Bangalore
Output: {"name":"Priya","salary":"60000","city":"Bangalore"}

Input: Ramesh,45000,Mumbai
Output: {"name":"Ramesh","salary":"45000","city":"Mumbai"}

Input: Neha,55000,Delhi
Output: {"name":"Neha","salary":"55000","city":"Delhi"}

"""
few_shot_response = ask_llm(few_shot_prompt,temperature = 0.0)
print("=== Conversion of text to JSON ===")
print()
print(few_shot_response)
try:
  parse = json.loads(few_shot_response.strip())
  print("Successfully parsed JSON :")
  print(f"Name : {parse['name']},Salary : {parse['salary']},City : {parse['city']}")
except json.JSONDecodeError:
  print("Failed to parse JSON - MODEL added extra text ")
  print("Solution : add explicit instructions in the system prompt")
```
