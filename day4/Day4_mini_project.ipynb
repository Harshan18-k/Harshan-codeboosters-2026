```python
import pandas as pd
import matplotlib.pyplot as plt
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.functions import year,month,to_date,col,round as spark_round
import warnings
warnings.filterwarnings('ignore')

data = pd.read_csv('/content/drive/MyDrive/Summer_Internship_2026/large_sales_data.csv')
print("=== DATA LOADED SUCCESSFULLY ===")
```

```python
fig , ax = plt.subplots(figsize = (10,8))
plt.xlabel("REGION")
plt.ylabel("REVENUE")
plt.title("MONTHLY REVENUE ANALYSIS")

# Group data by region and sum revenue
region_revenue = data.groupby('region')['revenue'].sum().reset_index()

# Get unique regions
unique_regions = region_revenue['region'].unique()

# Define a list of colors. If the number of unique regions is greater than this list,
# the colors will repeat, but each *unique* region will have a consistent color assigned.
colors_palette = ['red', 'green', 'blue', 'yellow', 'purple', 'orange', 'cyan', 'magenta', 'lime', 'pink', 'brown', 'grey']

# Create a dictionary to map each unique region to a color
region_color_map = {}
for i, region in enumerate(unique_regions):
    region_color_map[region] = colors_palette[i % len(colors_palette)]

# Create a list of colors corresponding to the regions in the grouped_data for plotting
bar_colors = [region_color_map[region] for region in region_revenue['region']]

# Plot the aggregated data with assigned colors
plt.barh(region_revenue['region'], region_revenue['revenue'], color=bar_colors)
plt.show()
```

```python
fig , ax = plt.subplots(figsize = (10,8))
plt.xlabel("REGION")
plt.ylabel("REVENUE")
plt.title("MONTHLY REVENUE ANALYSIS")

region_revenue = data.groupby('region')['revenue'].sum().reset_index()

unique_regions = region_revenue['region'].unique()

colors_palette = ['red', 'green', 'blue', 'yellow', 'purple', 'orange', 'cyan', 'magenta', 'lime', 'pink', 'brown', 'grey']

region_color_map = {}
for i, region in enumerate(unique_regions):
    region_color_map[region] = colors_palette[i % len(colors_palette)]


bar_colors = [region_color_map[region] for region in region_revenue['region']]


bars = plt.bar(region_revenue['region'], region_revenue['revenue'], color=bar_colors)
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 0.5, round(yval, 2), ha='center', va='bottom')
plt.show()
```
