```python
def run_query(query, db_name='internship.db'):
  conn = sqlite3.connect(db_name)
  data = pd.read_sql(query, conn)
  conn.close()
  print(f"=== Data from {db_name} Loaded Successfully ===")
  print("")
  return data
```

```python
query = "SELECT name , department  FROM internship WHERE department = 'Computer Science' "
print(run_query(query))
```

```python
query = "SELECT student_id , name FROM internship LIMIT 5"
print(run_query(query))
```
